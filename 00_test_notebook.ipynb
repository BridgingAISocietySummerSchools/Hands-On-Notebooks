### 🐍 Check Python Version

```python
import sys
print("Python version:", sys.version)

expected_major, expected_minor = 3, 13

# ✅ Check major + minor version only (patch version may vary)
major, minor = sys.version_info[:2]
assert (major, minor) == (expected_major, expected_minor), \
    f"Expected Python {expected_major}.{expected_minor}, but found {major}.{minor}."
```

### 📦 Check Package Versions (Based on requirements.txt)

```python
import requests
import re

def get_required_versions_from_text(txt):
    required_versions = {}
    for line in txt.splitlines():
        line = line.strip()
        if not line or line.startswith("#"):
            continue
        if any(op in line for op in ["<", ">", "=", "~"]):
            package = re.split(r"[<>=~]", line)[0]
            # 'notebook' is a local-only dependency, and 'ipywidgets' legitimately
            # differs between Colab and a local install -- see the separate check below.
            if package in ("notebook", "ipywidgets"):
                continue
            match = re.search(r"\d+\.\d+", line)
            if match:
                required_versions[package] = match.group()
    return required_versions

# 🔗 Load requirements.txt from GitHub
url = "https://raw.githubusercontent.com/BridgingAISocietySummerSchools/Hands-On-Notebooks/refs/heads/main/requirements.txt"
response = requests.get(url)
requirements_txt = response.text

required_versions = get_required_versions_from_text(requirements_txt)
```

```python
import importlib.metadata as metadata

for pkg, required in required_versions.items():
    version = metadata.version(pkg)
    print(f"{pkg}: {version}")
    assert version.startswith(required), f"Expected {pkg} version to start with {required}"

# ipywidgets is checked separately, because the right version depends on where
# you are running. Colab ships 7.x and draws the widgets with its own manager;
# a local Jupyter Notebook 7 / JupyterLab 4 needs 8.x, since ipywidgets 7 fails
# to render there and the sliders simply never appear. Both work for this course.
import ipywidgets

ipywidgets_major = int(ipywidgets.__version__.split(".")[0])
print(f"ipywidgets: {ipywidgets.__version__}")
assert ipywidgets_major in (7, 8), (
    f"Expected ipywidgets 7.x (Colab) or 8.x (local), found {ipywidgets.__version__}"
)
if ipywidgets_major == 7:
    print("   ℹ️  ipywidgets 7 detected. Fine on Colab; locally you want 8.x,")
    print("      otherwise the interactive sliders will not show up.")
```
